# TASK 3(Employee class)

```python
class Employee:
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

    def display(self):
        print(f"Employee ID: {self.emp_id}")
        print(f"Employee Name: {self.name}")
        print(f"Salary: {self.salary}\n")

emp1 = Employee(101, "Rahul", 50000)
emp2 = Employee(102, "Anjali", 65000)


emp1.display()
emp2.display()
```

```python
import numpy as np
import pandas as pd
 
np.random.seed(42)
 
rows = 1000
 
df = pd.DataFrame({
"Student_ID": range(1, rows + 1),
"Age": np.random.randint(18, 26, rows),
"Study_Hours": np.random.randint(1, 11, rows),
"Attendance": np.random.randint(60, 101, rows),
"Assignments": np.random.randint(1, 11, rows),
"Previous_Marks": np.random.randint(40, 101, rows),
"Sleep_Hours": np.random.randint(4, 10, rows),
"Internet_Usage": np.random.randint(1, 9, rows),
"Marks": np.random.randint(35, 101, rows)
})
 
print(df.head())
 
df.to_csv("student_performance_1000.csv", index=False)

print(df.head())
print(df.tail())
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
```

# Probability

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("student_performance_1000.csv")

# Students with attendance >= 90
high_attendance = df[df["Attendance"] >= 90]

# Students with marks > 80 among them
high_marks = high_attendance[high_attendance["Marks"] > 80]

# Probability
probability = len(high_marks) / len(high_attendance)

print("Probability of scoring >80 given Attendance ≥90:", probability)
```

# Class and Objects

```python
class Student:
    def __init__(self):
        self.id = int(input("Enter ID: "))
        self.name = input("Enter your name: ")
        self.studyhour = int(input("Enter study hours: "))
        self.attendance = input("Enter attendance: ")

    def show_details(self):
        print("ID:", self.id)
        print("Name:", self.name)
        print("Study Hours:", self.studyhour)
        print("Attendance:", self.attendance)

# Create object
stu1 = Student()

# Show details
stu1.show_details()
```

# Correlation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("student_performance_1000.csv")

# Compute correlation matrix
corr_matrix = df.corr(numeric_only=True)

# Plot heatmap
plt.figure(figsize=(10,8))
sns.heatmap(
    corr_matrix,
    annot=True,        
    cmap="coolwarm",  
    fmt=".2f",      
    linewidths=0.5     
)
plt.title("Correlation Matrix Heatmap")
plt.show()
```

# Loops

```python
import pandas as pd
df = pd.read_csv("student_performance_1000.csv")
#Print students with attendance > 90
print("Students with Attendance > 90:\n")
for i in range(len(df)):
    if df.loc[i, "Attendance"] > 90:
        print("ID:", df.loc[i, "Student_ID"], "| Attendance:", df.loc[i, "Attendance"])
```
