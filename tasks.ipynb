# TASK 3(Employee class)

```python
class Employee:
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

    def display(self):
        print(f"Employee ID: {self.emp_id}")
        print(f"Employee Name: {self.name}")
        print(f"Salary: {self.salary}\n")

emp1 = Employee(101, "Rahul", 50000)
emp2 = Employee(102, "Anjali", 65000)


emp1.display()
emp2.display()
```

# Probability

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("student_performance_1000.csv")

# Students with attendance >= 90
high_attendance = df[df["Attendance"] >= 90]

# Students with marks > 80 among them
high_marks = high_attendance[high_attendance["Marks"] > 80]

# Probability
probability = len(high_marks) / len(high_attendance)

print("Probability of scoring >80 given Attendance ≥90:", probability)
```

# Class and Objects

```python
class Student:
    def __init__(self):
        self.id = int(input("Enter ID: "))
        self.name = input("Enter your name: ")
        self.studyhour = int(input("Enter study hours: "))
        self.attendance = input("Enter attendance: ")

    def show_details(self):
        print("ID:", self.id)
        print("Name:", self.name)
        print("Study Hours:", self.studyhour)
        print("Attendance:", self.attendance)

# Create object
stu1 = Student()

# Show details
stu1.show_details()
```

# Correlation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("student_performance_1000.csv")

# Compute correlation matrix
corr_matrix = df.corr(numeric_only=True)

# Plot heatmap
plt.figure(figsize=(10,8))
sns.heatmap(
    corr_matrix,
    annot=True,        
    cmap="coolwarm",  
    fmt=".2f",      
    linewidths=0.5     
)
plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python
import pandas as pd
df = pd.read_excel(r'C:\\Users\\hp\\OneDrive\Documents\\Desktop\\data.xlsx')
df
```

```python
import matplotlib.pyplot as plt
avg_salary = df.groupby("JobCity")['Salary'].mean().sort_values(ascending=False)
top_cities = avg_salary.head(10)   # only top 10
plt.figure(figsize=(10,6))
plt.bar(top_cities.index.astype(str),
      top_cities.values,
        color=plt.cm.tab10(range(len(top_cities))))
plt.title("Average of Salary by JobCity")
plt.xlabel('JobCity')
plt.ylabel('Average Salary')
plt.tight_layout()
plt.show()
```

# Pie Chart

```python
import matplotlib.pyplot as plt
avg_salary = df.groupby("JobCity")['Salary'].mean().sort_values(ascending=False)
top_cities = avg_salary.head(10)

plt.figure(figsize=(7,7))
plt.pie(top_cities.values,
        labels=top_cities.index.astype(str),
        autopct="%1.1f%%",
        startangle=90)

plt.title("Average Salary by Top 10 JobCities")
plt.show()
```

# Degree vs Salary

```python
import matplotlib.pyplot as plt
import numpy as np

degree_cols = ["ComputerScience","MechanicalEngg","ElectricalEngg","TelecomEngg","CivilEngg"]
degree_salary = {col: df.loc[df[col] != 0, "Salary"].mean()
                 for col in degree_cols
                 if pd.notna(df.loc[df[col] != 0, "Salary"].mean())}

plt.figure(figsize=(8,6))
plt.bar(degree_salary.keys(),
        degree_salary.values(),
        color=plt.cm.tab10(range(len(degree_salary))))
plt.title("Mean Salary by Degree")
plt.xlabel("Degree")
plt.ylabel("Mean Salary")
plt.tight_layout()
plt.show()
```
