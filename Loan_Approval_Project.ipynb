# Data Cleaning

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv(r"C:\Users\hp\OneDrive\Documents\Desktop\ML\datasets\training.csv")
df = df.drop("Loan_ID", axis=1)
print(df.head())
```

```python
#HANDLE CATEGORICAL MISSING VALUES
categorical_columns = [
    "Gender",
    "Married",
    "Dependents",
    "Self_Employed"
]

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
print("Categorical missing values handled.")
```

```python
#HANDLE NUMERICAL MISSING VALUES
numerical_columns = [
    "LoanAmount",
    "Loan_Amount_Term",
    "Credit_History"
]

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())
print("Numerical missing values handled.")
```

```python
print("Missing Values After Preprocessing:")
print(df.isnull().sum())
```

```python
# CONVERT DEPENDENTS
df["Dependents"] = df["Dependents"].replace("3+", "3")

# Convert Dependents to integer
df["Dependents"] = df["Dependents"].astype(int)

print("Unique Dependents values:")
print(df["Dependents"].unique())
```

# Data Preprocessing

```python
# LABEL ENCODING
from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder
le = LabelEncoder()

categorical_columns = [
    "Gender",
    "Married",
    "Education",
    "Self_Employed",
    "Property_Area",
    "Loan_Status"
]

# Encode categorical columns
for col in categorical_columns:
    df[col] = le.fit_transform(df[col])

print("Categorical encoding completed!")
print(df.head())
```

```python
# CHECK DATA TYPES
print("Data Types After Encoding:")
print(df.dtypes)
```

```python
#Feature Engineering

df["TotalIncome"] = (
    df["ApplicantIncome"] +
    df["CoapplicantIncome"]
)

print("TotalIncome created successfully!")

print(
    df[
        ["ApplicantIncome",
         "CoapplicantIncome",
         "TotalIncome"]
    ].head()
)
```

```python
#SEPARATE FEATURES AND TARGET
X = df.drop("Loan_Status", axis=1)

y = df["Loan_Status"]

print("Features (X):")
print(X.head())

print("\nTarget (y):")
print(y.head())
```

```python
# training and testing data
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
#Feature Scaling
from sklearn.preprocessing import StandardScaler

# Create scaler
scaler = StandardScaler()

# Fit ONLY on training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform test data using the same scaler
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed successfully!")

print("Scaled Training Data Shape:", X_train_scaled.shape)
print("Scaled Testing Data Shape:", X_test_scaled.shape)
```

# Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    mean_squared_error,
    r2_score,
    confusion_matrix
)

import matplotlib.pyplot as plt
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
y_pred_linear = linear_model.predict(X_test)
y_pred_linear_class = (y_pred_linear >= 0.5).astype(int)
linear_accuracy = accuracy_score(
    y_test,
    y_pred_linear_class
)

linear_precision = precision_score(
    y_test,
    y_pred_linear_class,
    zero_division=0
)

linear_recall = recall_score(
    y_test,
    y_pred_linear_class,
    zero_division=0
)

linear_f1 = f1_score(
    y_test,
    y_pred_linear_class,
    zero_division=0
)

linear_mse = mean_squared_error(
    y_test,
    y_pred_linear
)

linear_r2 = r2_score(
    y_test,
    y_pred_linear
)

print("LINEAR REGRESSION RESULTS")
print("=========================")

print("Accuracy :", round(linear_accuracy, 4))
print("Precision:", round(linear_precision, 4))
print("Recall   :", round(linear_recall, 4))
print("F1 Score :", round(linear_f1, 4))
print("MSE      :", round(linear_mse, 4))
print("R2 Score :", round(linear_r2, 4))


# Confusion Matrix
cm_linear = confusion_matrix(
    y_test,
    y_pred_linear_class
)

print("\nConfusion Matrix:")
print(cm_linear)

plt.figure(figsize=(8, 5))

plt.scatter(
    range(len(y_test)),
    y_test,
    label="Actual",
    color="blue",
    marker="o"
)

plt.scatter(
    range(len(y_pred_linear_class)),
    y_pred_linear_class,
    label="Predicted",
    color="red",
    marker="x"
)

plt.title("Linear Regression - Actual vs Predicted")
plt.xlabel("Applicant Record Number")
plt.ylabel("Loan Approval Status")

plt.yticks(
    [0, 1],
    ["Not Approved", "Approved"]
)

plt.legend()

plt.show()
```

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

logistic_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)
# Parameters for tuning
param_grid = {
    "C": [0.01, 0.1, 1, 10, 100],
    "solver": ["liblinear", "lbfgs"]
}

# Find the best model internally
grid_logistic = GridSearchCV(
    estimator=logistic_model,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy"
)

grid_logistic.fit(
    X_train_scaled,
    y_train
)
# Use the best model
logistic_model = grid_logistic.best_estimator_
# Make predictions
y_pred_logistic = logistic_model.predict(
    X_test_scaled
)
# Evaluation metrics
logistic_accuracy = accuracy_score(
    y_test,
    y_pred_logistic
)

logistic_precision = precision_score(
    y_test,
    y_pred_logistic,
    zero_division=0
)

logistic_recall = recall_score(
    y_test,
    y_pred_logistic,
    zero_division=0
)

logistic_f1 = f1_score(
    y_test,
    y_pred_logistic,
    zero_division=0
)
# Display results
print("LOGISTIC REGRESSION RESULTS")
print("===========================")

print("Accuracy :", round(logistic_accuracy, 4))
print("Precision:", round(logistic_precision, 4))
print("Recall   :", round(logistic_recall, 4))
print("F1 Score :", round(logistic_f1, 4))


# Confusion Matrix
cm_logistic = confusion_matrix(
    y_test,
    y_pred_logistic
)

print("\nConfusion Matrix:")
print(cm_logistic)
```

# K-Nearest Neighbors (KNN)

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
# Create model
knn_model = KNeighborsClassifier()

# Parameters for tuning
param_grid = {
    "n_neighbors": range(3, 21, 2),
    "weights": ["uniform", "distance"],
    "metric": ["euclidean", "manhattan"]
}
# Find the best model internally
grid_knn = GridSearchCV(
    estimator=knn_model,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy"
)

grid_knn.fit(
    X_train_scaled,
    y_train
)
# Use the best model
knn_model = grid_knn.best_estimator_

# Make predictions
y_pred_knn = knn_model.predict(
    X_test_scaled
)
# Evaluation metrics
knn_accuracy = accuracy_score(
    y_test,
    y_pred_knn
)

knn_precision = precision_score(
    y_test,
    y_pred_knn,
    zero_division=0
)

knn_recall = recall_score(
    y_test,
    y_pred_knn,
    zero_division=0
)

knn_f1 = f1_score(
    y_test,
    y_pred_knn,
    zero_division=0
)
# Display results
print("K-NEAREST NEIGHBORS (KNN) RESULTS")
print("=================================")

print("Accuracy :", round(knn_accuracy, 4))
print("Precision:", round(knn_precision, 4))
print("Recall   :", round(knn_recall, 4))
print("F1 Score :", round(knn_f1, 4))

# Confusion Matrix
cm_knn = confusion_matrix(
    y_test,
    y_pred_knn
)

print("\nConfusion Matrix:")
print(cm_knn)
```

# DECISION TREE CLASSIFICATION

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Create model
tree_model = DecisionTreeClassifier(
    random_state=42
)

# Parameters for tuning
param_grid = {
    "criterion": ["gini", "entropy"],
    "max_depth": [3, 4, 5, 6, 8, 10],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4]
}

# Find the best model internally
grid_tree = GridSearchCV(
    estimator=tree_model,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy"
)

grid_tree.fit(
    X_train,
    y_train
)

# Use the best model
tree_model = grid_tree.best_estimator_

# Make predictions
y_pred_tree = tree_model.predict(
    X_test
)

# Evaluation metrics
tree_accuracy = accuracy_score(
    y_test,
    y_pred_tree
)

tree_precision = precision_score(
    y_test,
    y_pred_tree,
    zero_division=0
)

tree_recall = recall_score(
    y_test,
    y_pred_tree,
    zero_division=0
)

tree_f1 = f1_score(
    y_test,
    y_pred_tree,
    zero_division=0
)

# Display results
print("DECISION TREE RESULTS")
print("=====================")

print("Accuracy :", round(tree_accuracy, 4))
print("Precision:", round(tree_precision, 4))
print("Recall   :", round(tree_recall, 4))
print("F1 Score :", round(tree_f1, 4))


# Confusion Matrix
cm_tree = confusion_matrix(
    y_test,
    y_pred_tree
)

print("\nConfusion Matrix:")
print(cm_tree)
```

# Model Comparison

```python
import pandas as pd

final_results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Logistic Regression",
        "KNN",
        "Decision Tree"
    ],

    "Accuracy": [
        linear_accuracy,
        logistic_accuracy,
        knn_accuracy,
        tree_accuracy
    ],

    "Precision": [
        linear_precision,
        logistic_precision,
        knn_precision,
        tree_precision
    ],

    "Recall": [
        linear_recall,
        logistic_recall,
        knn_recall,
        tree_recall
    ],

    "F1 Score": [
        linear_f1,
        logistic_f1,
        knn_f1,
        tree_f1
    ]
})

# Round the values for easy reading
final_results = final_results.round(4)

# Display final comparison
print("FINAL MODEL COMPARISON")
print("======================")

print(final_results.to_string(index=False))
```

```python
# BEST-PERFORMING MODEL

# Find the model with the highest F1 Score
best_model = final_results.loc[
    final_results["F1 Score"].idxmax(),
    "Model"
]

best_f1_score = final_results["F1 Score"].max()
print("BEST-PERFORMING MODEL")
print("=====================")

print("Model    :", best_model)
print("F1 Score :", best_f1_score)

best_model_result = final_results[
    final_results["Model"] == best_model
]

print("\nPerformance of Best Model:")
print(best_model_result.to_string(index=False))
```
