```python
import numpy as np

# Generate random marks with mean=70, std=15, for 500 students
marks = np.random.normal(loc=70, scale=15, size=500)

# Clip values so they stay between 0 and 100
marks = np.clip(marks, 0, 100)

# Count students in different ranges
between_60_80 = np.sum((marks >= 60) & (marks <= 80))
above_90 = np.sum(marks > 90)
below_50 = np.sum(marks < 50)

# Print results
print("Students scoring between 60 and 80:", between_60_80)
print("Students scoring above 90:", above_90)
print("Students scoring below 50:", below_50)
```

# OUTLIERS

```python
import pandas as pd
df = pd.read_excel(r'C:\\Users\\hp\\OneDrive\Documents\\Desktop\\data.xlsx')
df
```

```python
print("Dataset Shape:", df.shape)
print("\n Salary statistics before outlier treatment:")
print(df['Salary'].describe())
```

```python
q1 = df['Salary'].quantile(0.25)
q3 = df['Salary'].quantile(0.75)

IQR = q3 - q1

lower_limit = q1 - 1.5 * IQR
upper_limit = q3 + 1.5 * IQR

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", IQR)
print("Lower:", lower_limit)
print("Upper:", upper_limit)

outliers = df[
    (df["Salary"] < lower_limit) |
    (df["Salary"] > upper_limit)
]

print("\nNumber of Salary Outliers:", len(outliers))

print("\nSalary Outliers:")
print(outliers[["ID", "Salary"]].head(20))

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.boxplot(df['Salary'])
plt.title('Salary Before Outlier Treatment')
plt.ylabel('Salary')   # <-- function call, not assignment
plt.show()
```

```python
df['Salary_Capped'] = df['Salary'].clip(
    lower=lower_limit,
    upper=upper_limit)
```

```python

print("After Outliers")

print("Original Minimum Salary:", df["Salary"].min())
print("Original Maximum Salary:", df["Salary"].max())

print("\nCapped Minimum Salary:", df["Salary_Capped"].min())
print("Capped Maximum Salary:", df["Salary_Capped"].max())

print("\nOriginal Mean Salary:", df["Salary"].mean())
print("Capped Mean Salary:", df["Salary_Capped"].mean())

print("\nOriginal Median Salary:", df["Salary"].median())
print("Capped Median Salary:", df["Salary_Capped"].median())
```

```python
plt.figure(figsize=(8,5))
plt.boxplot(df['Salary_Capped'])
plt.title("Salary after Outlier Treatment")
plt.ylabel('Salary_Capped')
plt.show()
```
