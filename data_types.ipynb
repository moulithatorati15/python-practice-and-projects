```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
x = np.linspace(-4,4,100)
y = norm.pdf(x)
plt.plot(x,y)
plt.title("NORMAL DISTRIBUTION")
plt.xlabel("X Values")
plt.ylabel("Probability Density")
plt.show()
```

```python
import numpy as np

# Generate random marks with mean=70, std=15, for 500 students
marks = np.random.normal(loc=70, scale=15, size=500)

# Clip values so they stay between 0 and 100
marks = np.clip(marks, 0, 100)

# Count students in different ranges
between_60_80 = np.sum((marks >= 60) & (marks <= 80))
above_90 = np.sum(marks > 90)
below_50 = np.sum(marks < 50)

# Print results
print("Students scoring between 60 and 80:", between_60_80)
print("Students scoring above 90:", above_90)
print("Students scoring below 50:", below_50)
```
